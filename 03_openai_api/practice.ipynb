```python
from openai import OpenAI

from dotenv import load_dotenv
import os
load_dotenv()
api_key = os.getenv('openai_key')
client = OpenAI(api_key = api_key)

resp = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "토큰 사용량 테스트"}
    ]
)

print(resp.choices[0].message.content)  # 응답 내용
print(resp.usage)

import time
print(time.strftime("%Y-%m-%d %H:%M:%S"), resp.usage)
```
