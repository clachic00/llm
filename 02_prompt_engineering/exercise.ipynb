```python
from openai import OpenAI                 # OpenAI API를 사용하기 위한 클라이언트 클래스
from dotenv import load_dotenv            # .env 파일에서 환경변수를 불러오는 함수
import os                                 # 운영체제 환경변수 접근용 모듈

load_dotenv()                             # .env 파일을 읽어서 환경변수로 등록
client = OpenAI(api_key=os.getenv("openai_key"))  # 환경변수에 저장된 OpenAI API 키로 클라이언트 생성

# Responses API 텍스트 생성 요청
response = client.responses.create(
    model="gpt-4.1-mini",                 # 사용할 GPT 모델 지정
    input="AI 로봇에 대한 스토리 만들어줘.", # 모델에게 전달할 프롬프트(질문)
    max_output_tokens=200,                # 생성될 최대 토큰 수 제한
    top_p=0.9,                            # nucleus sampling 비율 (확률 상위 90%만 고려)
    temperature=0.7                      # 출력 다양성 조절 (낮을수록 보수적, 높을수록 창의적)
)

print(response.output_text)               # 모델이 생성한 최종 텍스트 출력
```
